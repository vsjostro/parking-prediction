```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file into a pandas DataFrame with correct delimiter
df = pd.read_csv('..\\preprocess\parking_time_series.csv', delimiter=',')

# Convert the 'Time' column to datetime format
df['Time'] = pd.to_datetime(df['Time'])

# Set 'Time' column as index
df.set_index('Time', inplace=True)

# Resample the DataFrame to 15-minute intervals and sum the values
df_15min_sum = df.resample('15T').sum()

# Plotting the time series data using a line plot
plt.figure(figsize=(10, 6))
plt.plot(df_15min_sum.index, df_15min_sum['Parkings'], marker='o', linestyle='-')
plt.xlabel('Time')
plt.ylabel('Total amount of parkings')
plt.title('Total amount of parking over time (15-minute intervals)')
plt.grid(True)
plt.show()
```
